# Session 11 walkthrough

This notebook is a thin front-end on the library in `src/`. Experiments are run with `python -m run_experiments --overwrite`. Measured numbers live in `artifacts/results.json` and are discussed in `report/final_report.md`.

```python
import json
import sys
from pathlib import Path

ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))

from adam_manual import verify_against_pytorch
from utils import load_config

cfg = load_config(ROOT / "configs" / "default.yaml")
adam = verify_against_pytorch(**{k: cfg["adam"][k] for k in ("gradients", "w0", "lr", "beta1", "beta2", "eps")})
print("worst abs error vs torch", adam["manual_vs_torch"]["worst_abs_error"])
print("agrees", adam["agrees_with_torch"])
print("uncorrected m_hat at step 1", adam["manual_uncorrected"][0]["m_hat"])
```

```python
results_path = ROOT / "artifacts" / "results.json"
if results_path.exists():
    results = json.loads(results_path.read_text())
    print("tasks", results.get("tasks"))
    if "B" in results:
        sc = results["B"]["scenarios"]["constant"]
        print("constant rel-diff step1", sc["first_step_relative_difference"])
        print("constant rel-diff step20", sc["step20_relative_difference"])
        print("agree by 20?", sc["agree_by_step20"])
        print("persistent crossing", sc["persistent_crossing"])
    if "D" in results:
        print("val@200 cosine", results["D"]["cosine_300"]["step200"])
        print("val@200 wsd", results["D"]["wsd_300"]["step200"])
    if "E" in results:
        print("extrapolation", results["E"]["extrapolation"])
else:
    print("Run python -m run_experiments --overwrite to populate artifacts.")
```
